```python
# Queue implementation using List

queue = []
```

```python
# Enqueue - add elements
queue.append(10)
queue.append(20)
queue.append(30)
```

```python
print("Queue:", queue)
```

```python
# Dequeue - remove from front
removed = queue.pop(0)
print("Removed:", removed)
```
